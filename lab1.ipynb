```python
#Data Type Identification
def identify_data_type(user_input):
    try:
        return int(user_input), "int"
    except ValueError:
        pass

    try:
        return float(user_input), "float"
    except ValueError:
        pass

    if user_input.lower() == 'true' or user_input.lower() == 'false':
        return user_input.lower() == 'true', "bool"

    return user_input, "str"

user_input = input("Enter something: ")

value, data_type = identify_data_type(user_input)
print(f"Value: {value}, Data Type: {data_type}")
```

```python
#Palindrome Checker
def is_palindrome(s):
    # Normalize the string: remove spaces and convert to lowercase
    normalized_string = ''.join(s.split()).lower()

    # Lambda function to check if the normalized string is a palindrome
    check_palindrome = lambda x: x == x[::-1]

    # Use the lambda function and return the result
    return check_palindrome(normalized_string)

input_string = input("Enter a string to check if it's a palindrome: ")

if is_palindrome(input_string):
    print(f'"{input_string}" is a palindrome.')
else:
    print(f'"{input_string}" is not a palindrome.')
```

```python
#Fibonacci Sequence Generator
def Fibonacci_series(num):
    if num <= 0:
        return []
    elif num == 1:
        return [0]
    elif num == 2:
        return [0, 1]
    else:
        fibonacci = [0, 1]
        for i in range(2, num):
            next_number = fibonacci[i - 1] + fibonacci[i - 2]
            fibonacci.append(next_number)
        return fibonacci

def main():
 num = int(input("Enter the number of Fibonacci numbers to generate: "))
 fibonacci_series = Fibonacci_series(num)
 print(fibonacci_series)

if __name__ == "__main__":
    main()
```

```python
#Average Calculator with Validation
def Input_Validation():
    while True:
        value = input("Enter a number or -1 to stop: ")
        try:
            value = float(value)  # Convert input to float
            if value == -1:  # Check for -1 to stop the loop
                return value
            elif value < 0:
                print("Please enter a non-negative number.")
            else:
                return value  # Valid
        except ValueError:
            print("Invalid input. Please enter a number.")

def Average(numbers):
    return sum(numbers) / len(numbers)

def main():
    numbers = []
    while True:
        num = Input_Validation()
        if num == -1:  # Stop if -1
            break
        numbers.append(num)

    if numbers:  # list is not empty
        average = Average(numbers)
        print(f"Average: {average:.2f}")  # Format the average to 2 decimal places
    else:
        print("No numbers entered.")

if __name__ == "__main__":
    main()
```

```python
# User registration and login system
users = {}  # Dictionary to store user credentials

def register():
    username = input("Enter a username: ")
    if username in users:
        print("Username already exists! Please choose a different one.")
        return

    password = input("Enter a password: ")
    # Store the user's credentials in the dictionary
    users[username] = password
    print("Registration successful!")

def login():
    username = input("Enter your username: ")
    if username not in users:
        print("Username not found!")
        return

    password = input("Enter your password: ")
    if users[username] == password:
        print("Login successful!")
    else:
        print("Incorrect password!")

def main():
    while True:
        print("\n--- User Registration System ---")
        print("1. Register")
        print("2. Login")
        print("3. Exit")
        choice = input("Choose an option (1-3): ")

        if choice == '1':
            register()
        elif choice == '2':
            login()
        elif choice == '3':
            print("Exiting the system. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    main()
```

```python
#Counting Elements with a Dictionary
def count_word_frequency():
    words_input = input("Enter a list of words separated by spaces: ")
    words = words_input.split()
    # Dictionary to store word frequencies
    frequency = {}

    # Counting the frequency of each word
    for word in words:
    # Convert to lowercase to make the counting case-insensitive
        word = word.lower()
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    print("\nWord Frequency:")
    for word, count in frequency.items():
        print(f"{word}: {count}")

count_word_frequency()
```

```python
#Temperature Converter
def convert_temperature():
    print("Temperature Conversion")
    print("1. Celsius to Fahrenheit")
    print("2. Fahrenheit to Celsius")

    choice = input("Choose an option (1 or 2): ")

    if choice == '1':
        # Celsius to Fahrenheit conversion
        celsius = float(input("Enter temperature in Celsius: "))
        fahrenheit = (celsius * 9/5) + 32
        print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")

    elif choice == '2':
        # Fahrenheit to Celsius conversion
        fahrenheit = float(input("Enter temperature in Fahrenheit: "))
        celsius = (fahrenheit - 32) * 5/9
        print(f"{fahrenheit}°F is equal to {celsius:.2f}°C")

    else:
        print("Invalid choice! Please select either 1 or 2.")

convert_temperature()
```
